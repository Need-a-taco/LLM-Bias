```python
import pandas as pd
from collections import Counter, defaultdict
import re
import matplotlib.pyplot as plt
import seaborn as sns
import math
```

```python
df = pd.read_csv("../data/papers.csv")
df.columns
```

Focus on statistics related to data access.

```python
applications_lst = (df["Data Domain"].tolist())
len(applications_lst)
```

```python
applications_freqs = defaultdict(int)

for a in applications_lst:  

    if isinstance(a, float):
        continue

    s = a.strip().lower()
    
    applications = [l.strip() for l in s.split(',')]
    
    for application in applications:
        if "other" in application or ")" in application:
            applications_freqs["Other"] += 1
        else:
            applications_freqs[application] += 1

applications_freqs = dict(sorted(applications_freqs.items(), key=lambda kv: kv[1]))
applications_freqs = dict(reversed(list(applications_freqs.items())))


```

```python
label_rotation = 0  # degrees; tilts y-axis category labels (use 0 for horizontal text)
x_tick_labelsize = 18  # tick label font size on both axes

# Axis titles ("Frequency", "Task Type"): False hides both; or comment out the block below.
SHOW_AXIS_TITLES = True
axis_title_fontsize = 22

# Horizontal gap between bar end and count label (x-axis data units); increase for more space.
count_label_pad = 0.35

# Space (points) between the rotated y-axis title ("Task Type") and the category labels; increase to push title left.
y_axis_title_labelpad = 20

# Write vector output next to this notebook when True (run with cwd = medical_applications/).
SAVE_SVG = True
svg_path = "data_domain.svg"

applications = list(applications_freqs.keys())
counts = list(applications_freqs.values())
plot_df = pd.DataFrame(
    {"task_type": [a.title().replace("Ehr", "EHR") for a in applications], "count": counts}
)

n_cats = len(applications)
fig, ax = plt.subplots(figsize=(12, max(8, 0.38 * n_cats)))

sns.barplot(
    data=plot_df,
    y="task_type",
    x="count",
    ax=ax,
    color="royalblue",
    edgecolor="black",
    linewidth=1,
    order=plot_df["task_type"],
)

sns.despine(ax=ax)

xmax = max(counts) if counts else 1
x_gap = count_label_pad
x_right = xmax + x_gap + max(0.7, 0.07 * xmax)  # room so labels are not clipped

for i, count in enumerate(counts):
    ax.text(count + x_gap, i, f"{count}", ha="left", va="center")

ax.set_xlim(0, x_right)
tick_hi = max(5, math.ceil(x_right / 5) * 5)
ax.set_xticks(range(0, tick_hi + 1, 5))

# --- axis titles (easy to disable: set SHOW_AXIS_TITLES = False above, or comment out this block) ---
if SHOW_AXIS_TITLES:
    ax.set_xlabel("Frequency", fontsize=axis_title_fontsize)
    ax.set_ylabel("Task Type", fontsize=axis_title_fontsize, labelpad=y_axis_title_labelpad)

ax.tick_params(axis="x", labelsize=x_tick_labelsize)
ax.tick_params(axis="y", labelsize=x_tick_labelsize, pad=2)
ax.set_yticklabels(ax.get_yticklabels(), rotation=label_rotation, ha="right")
# Manual margins: avoid over-wide left gutter (constrained layout often over-reserves)
longest = max((len(str(t.get_text())) for t in ax.get_yticklabels()), default=10)
fig.subplots_adjust(
    left=min(0.38, max(0.11, 0.085 + 0.0058 * longest)),
    right=0.98,
    top=0.98,
    bottom=0.08,
)
if SAVE_SVG:
    fig.savefig(svg_path, format="svg", bbox_inches="tight")
plt.show()
```
