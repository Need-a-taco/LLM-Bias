```python
import pandas as pd

df = pd.read_csv("data/papers.csv")
df.head()
```

```python
# Gender labels in Conclusions are expected in a balanced form (Female + Male).
# Papers that mention Female but not Male (or vice versa) break that pattern.

conclusions = df["Conclusions"].astype(str)

female_not_male = df[
    conclusions.str.contains("Female", na=False)
    & ~conclusions.str.contains("Male", na=False)
].copy()

male_not_female = df[
    conclusions.str.contains("Male", na=False)
    & ~conclusions.str.contains("Female", na=False)
].copy()

print(f"Papers with 'Female' in Conclusions but not 'Male': {len(female_not_male)}")
print(f"Papers with 'Male' in Conclusions but not 'Female': {len(male_not_female)}")
```
